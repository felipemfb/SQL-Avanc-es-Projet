# Exploratory Data Analysis

## Descriptive statistic for each file

### Importing all the libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Reading the databases

```python
dbNames = ['characters', 'directors', 'genres', 'knownformovies', 'movies', 'persons', 'principals', 'professions', 'ratings', 'titles', 'writers']
db = []
for name in dbNames:
    df = pd.read_csv(f"csv/{name}.csv")
    db.append((df))
```

### Printing the headers

```python
for name, df in zip(dbNames, db):
    print(f"Dataset name: {name}")
    print(df.head())
    print("\n")
```

### Printing the number of lines and columns

```python
for name, df in zip(dbNames, db):
    print(f"Dataset name: {name}")
    rows, columns = df.shape
    print(f"Rows: {rows}, Columns: {columns}")
    print("\n")
```

### Printing all the types of data

```python
for name, df in zip(dbNames, db):
    print(f"Dataset name: {name}")
    print(df.dtypes)
    print("\n")
```

### Printing the missing data as NaN and empty strings

```python
missing = []

for name, df in zip(dbNames, db):
    print(f"Dataset name: {name}")
    NaN = df.isna().sum()
    empty = (df.applymap(lambda x: isinstance(x, str) and x.strip() == "")).sum()
    t = NaN + empty
    missing.append(t)
    total = t.sum()
    print(f"{total} values are missing on this dataset\n")
    print("\n")
```

### Printing all the unique data by column

```python
for name, df in zip(dbNames, db):
    print(f"Dataset name: {name}")
    for col in df.columns:
        print(f"Column name: {col}")
        print(df[col].unique())
        print(f"Number of unique values: {len(df[col].unique())}")
        print("\n")

```

## Exploratory Data Analysis

### Histogram of Films Distribution by Year

```python
i = dbNames.index('movies')
j = db[i]["('startYear',)"]

plt.figure(figsize=(8,5))
j.hist(bins = 100)
plt.xlabel('Start Year')
plt.ylabel('Count')
plt.title('Start Year vs Count Histogram')
plt.xlim(j.min(), j.max())
plt.show()
```

### Top 10 genders plus frequents

```python
i = dbNames.index('genres')
j = db[i]["('genre',)"]

bins = range(0,10)
plt.figure(figsize=(8,5))
plt.xlim(0, 9)
j.hist(bins = bins)
plt.xticks(bins)
plt.xlabel('Gender Types')
plt.ylabel('Count')
plt.title('Gender Types vs Count Histogram')
plt.show()
```

### Ratings distribution

```python
i = dbNames.index('ratings')
j = db[i]["('averageRating',)"]
bins = range(0,10)
plt.figure(figsize=(8,5))
j.hist(bins=bins)
plt.xticks(bins)
plt.xlim(0, 9)
plt.xlabel('Ratings')
plt.ylabel('Count')
plt.title('Ratings vs Count Histogram')
plt.show()
```

### Medium number of actors by film

```python
i = dbNames.index('knownformovies')
j = db[i][["('pid',)","('mid',)"]]
k = []

count = j.groupby("('mid',)")["('pid',)"].count()

bins = range(0, count.max(), 5)
plt.figure(figsize=(8,5))
count.hist(bins=bins)
plt.xlabel('Average number of actors')
plt.ylabel('Count')
plt.title('Ratings vs Count Histogram')
plt.xticks(bins)
plt.xlim(0, count.max() + 1)
plt.show()
```

## Tables relations

### Verifying the forgein keys that don't exist on the original table and identifying the orphan data

```python
movies = dbNames.index('movies')
mid = set(db[movies]["('mid',)"].astype(str))
persons = dbNames.index('persons')
pid = set(db[persons]["('pid',)"].astype(str))

print("===============================================================================================")
for name in dbNames:
    print(f"Analyzing the orphan data of the {name} dataframe...")
    missing = 0
    df = dbNames.index(name)
    if "('mid',)" in db[df].columns:
        for id in db[df]["('mid',)"]:
            if str(id) not in mid:
                missing += 1
                print (f"[Orphan Data] - Movie with id {id} at the table {name} not in registered at the table of movies")
        print(f"In total, {missing} movies of the table {name} are not registered at the table of movies")
        if "('pid',)" in db[df].columns:
            print("")
    missing = 0
    if "('pid',)" in db[df].columns:
        for id in db[df]["('pid',)"]:
            if str(id) not in pid:
                missing += 1
                print (f"[Orphan Data] - Person with id {id} at the table {name} not in registered at the table of persons")
        print(f"In total, {missing} persons of the table {name} are not registered at the table of persons")
    print("===============================================================================================")
```
